## Exercício 01 🟢 — O maior de dois números

Faça um programa que peça **dois números inteiros** e mostre qual é o maior. Se forem iguais, informe que são iguais.

**Exemplos de execução:**

*Exemplo 1:*

```text
Primeiro número: 7
Segundo número: 3
7 é o maior!
```

*Exemplo 2:*

```text
Primeiro número: 5
Segundo número: 5
Os números são iguais.
```

<details>
<summary>💡 <b>Dica</b> (clique para abrir)</summary>

Converta as entradas com `int()`! Comparando textos, `'10' > '9'` dá `False`, porque o Python compara caractere por caractere.

</details>

````python
# Exercício 01 — O maior de dois números
num1 = int(input("Primeiro número: "))
num2 = int(input("Segundo número: "))

if num1 > num2:
    print(num1, "é o maior!")
elif num2 > num1:
    print(num2, "é o maior!")
else:
    print("Os números são iguais.")
````

## Exercício 02 🟢 — Par ou ímpar

Faça um programa que peça um **número inteiro** e diga se ele é **par** ou **ímpar**.

**Exemplos de execução:**

*Exemplo 1:*

```text
Digite um número inteiro: 14
14 é par.
```

*Exemplo 2:*

```text
Digite um número inteiro: 7
7 é ímpar.
```

## Exercício 03 🟢 — Saudação por turno

Faça um programa que pergunte em que turno você estuda:

- **M** — Matutino
- **V** — Vespertino
- **N** — Noturno

Mostre `Bom Dia!`, `Boa Tarde!`, `Boa Noite!` ou `Valor Inválido!`, conforme o caso. O programa deve aceitar letras **maiúsculas e minúsculas**.

**Exemplos de execução:**

*Exemplo 1:*

```text
Em que turno você estuda? (M/V/N): v
Boa Tarde!
```

*Exemplo 2:*

```text
Em que turno você estuda? (M/V/N): x
Valor Inválido!
```

<details>
<summary>💡 <b>Dica</b> (clique para abrir)</summary>

Use `.upper()` na entrada para não precisar testar `'m'` e `'M'` separadamente.

</details>

````python
# Exercício 03 — Saudação por turno
turno = input("Em que turno você estuda? (M/V/N): ").upper()

if turno == "M":
    print("Bom Dia!")
elif turno == "V":
    print("Boa Tarde!")
elif turno == "N":
    print("Boa Noite!")
else:
    print("Valor Inválido!")
````

## Exercício 04 🟡 — Situação do aluno

Faça um programa que leia as **duas notas parciais** de um aluno, calcule a média e mostre:

- `Aprovado com Distinção`, se a média for igual a **10**;
- `Aprovado`, se a média for **maior ou igual a 7**;
- `Reprovado`, se a média for **menor que 7**.

**Exemplos de execução:**

*Exemplo 1:*

```text
Primeira nota: 8
Segunda nota: 6.5
Média: 7.25
Aprovado
```

*Exemplo 2:*

```text
Primeira nota: 10
Segunda nota: 10
Média: 10.0
Aprovado com Distinção
```

*Exemplo 3:*

```text
Primeira nota: 5
Segunda nota: 6
Média: 5.5
Reprovado
```

<details>
<summary>💡 <b>Dica</b> (clique para abrir)</summary>

A **ordem** das condições importa! Uma média 10 também é maior ou igual a 7. Qual condição deve ser testada primeiro?

</details>

````python
# Exercício 04 — Situação do aluno
nota1 = float(input("Primeira nota: "))
nota2 = float(input("Segunda nota: "))

media = (nota1 + nota2) / 2
print("Média:", media)

if media == 10:
    print("Aprovado com Distinção")
elif media >= 7:
    print("Aprovado")
else:
    print("Reprovado")
````

## Exercício 05 🟡 — Reajuste salarial

As Organizações Tabajara vão dar um aumento aos seus colaboradores. Faça um programa que receba o **salário atual** e aplique o reajuste conforme a tabela:

| Salário atual | Aumento |
|---|---|
| até R\$ 280,00 (inclusive) | 20% |
| acima de R\$ 280,00 até R\$ 700,00 (inclusive) | 15% |
| acima de R\$ 700,00 até R\$ 1500,00 (exclusive) | 10% |
| a partir de R\$ 1500,00 | 5% |

Mostre: o salário antes do reajuste, o percentual aplicado, o valor do aumento e o novo salário.

**Exemplos de execução:**

*Exemplo 1:*

```text
Salário atual: R$ 1000
Salário antes do reajuste: R$ 1000.00
Percentual aplicado: 10%
Valor do aumento: R$ 100.00
Novo salário: R$ 1100.00
```

*Exemplo 2:*

```text
Salário atual: R$ 250
Salário antes do reajuste: R$ 250.00
Percentual aplicado: 20%
Valor do aumento: R$ 50.00
Novo salário: R$ 300.00
```

<details>
<summary>💡 <b>Dica</b> (clique para abrir)</summary>

Guarde o percentual em uma variável dentro de cada `if/elif` e faça o cálculo **uma única vez**, depois da estrutura de decisão.

</details>

````python
# Exercício 05 — Reajuste salarial
salario = float(input("Salário atual: R$ "))

if salario <= 280:
    percentual = 20
elif salario <= 700:
    percentual = 15
elif salario < 1500:
    percentual = 10
else:
    percentual = 5

aumento = salario * (percentual / 100)
novo_salario = salario + aumento

print(f"Salário antes do reajuste: R$ {salario:.2f}")
print(f"Percentual aplicado: {percentual}%")
print(f"Valor do aumento: R$ {aumento:.2f}")
print(f"Novo salário: R$ {novo_salario:.2f}")
````

## Exercício 06 🟡 — Ano bissexto

Faça um programa que peça um **ano** e informe se ele é ou não **bissexto**.

Regra: um ano é bissexto se for **divisível por 4 e não for divisível por 100**, **ou** se for **divisível por 400**.

**Exemplos de execução:**

*Exemplo 1:*

```text
Digite um ano: 2024
2024 é bissexto.
```

*Exemplo 2:*

```text
Digite um ano: 1900
1900 não é bissexto.
```

*Exemplo 3:*

```text
Digite um ano: 2000
2000 é bissexto.
```

<details>
<summary>💡 <b>Dica</b> (clique para abrir)</summary>

Dá para resolver com um único `if`, combinando `and`, `or` e parênteses. Teste com 2024 (sim), 1900 (não) e 2000 (sim).

</details>

````python
# Exercício 06 — Ano bissexto
ano = int(input("Digite um ano: "))

if (ano % 4 == 0 and ano % 100 != 0) or (ano % 400 == 0):
    print(ano, "é bissexto.")
else:
    print(ano, "não é bissexto.")
````

## Exercício 07 🔴 — Classificador de triângulos

Faça um programa que peça os **3 lados** de um triângulo e informe se esses valores **podem formar um triângulo**. Se formarem, diga se ele é **equilátero**, **isósceles** ou **escaleno**.

- Três lados formam um triângulo quando a **soma de quaisquer dois lados é maior que o terceiro**;
- **Equilátero**: três lados iguais;
- **Isósceles**: exatamente dois lados iguais;
- **Escaleno**: três lados diferentes.

**Exemplos de execução:**

*Exemplo 1:*

```text
Lado A: 3
Lado B: 4
Lado C: 5
Triângulo escaleno.
```

*Exemplo 2:*

```text
Lado A: 5
Lado B: 5
Lado C: 8
Triângulo isósceles.
```

*Exemplo 3:*

```text
Lado A: 1
Lado B: 2
Lado C: 10
Esses valores não formam um triângulo.
```

<details>
<summary>💡 <b>Dica</b> (clique para abrir)</summary>

Primeiro verifique se é triângulo (três comparações ligadas com `and`). Só depois, *dentro* desse caso, classifique o tipo.

</details>

````python
# Exercício 07 — Classificador de triângulos
lado_a = float(input("Lado A: "))
lado_b = float(input("Lado B: "))
lado_c = float(input("Lado C: "))

if (lado_a + lado_b > lado_c) and (lado_a + lado_c > lado_b) and (lado_b + lado_c > lado_a):
    if lado_a == lado_b and lado_b == lado_c:
        print("Triângulo equilátero.")
    elif lado_a == lado_b or lado_a == lado_c or lado_b == lado_c:
        print("Triângulo isósceles.")
    else:
        print("Triângulo escaleno.")
else:
    print("Esses valores não formam um triângulo.")
````

## Exercício 08 🔴 — Equação do 2º grau

Faça um programa que calcule as raízes de uma equação do segundo grau, na forma **ax² + bx + c = 0**. O programa deve pedir os valores de `a`, `b` e `c` e tratar as seguintes situações:

- Se `a` for **zero**, a equação não é do 2º grau: informe e **não peça** os outros valores;
- Se o **delta** for **negativo**, a equação não possui raízes reais;
- Se o delta for **zero**, há apenas **uma raiz real**;
- Se o delta for **positivo**, há **duas raízes reais**.

Fórmulas: `Δ = b² − 4ac` e `x = (−b ± √Δ) / 2a`

**Exemplos de execução:**

*Exemplo 1:*

```text
Valor de a: 0
Com a = 0 a equação não é do segundo grau.
```

*Exemplo 2:*

```text
Valor de a: 1
Valor de b: -5
Valor de c: 6
Delta = 1.0
A equação possui duas raízes reais: 3.0 e 2.0
```

*Exemplo 3:*

```text
Valor de a: 1
Valor de b: 2
Valor de c: 1
Delta = 0.0
A equação possui uma raiz real: -1.0
```

*Exemplo 4:*

```text
Valor de a: 1
Valor de b: 0
Valor de c: 1
Delta = -4.0
A equação não possui raízes reais.
```

<details>
<summary>💡 <b>Dica</b> (clique para abrir)</summary>

Para a raiz quadrada use `math.sqrt(delta)` ou `delta ** 0.5`. No notebook, prefira organizar o código com `if/else` aninhados em vez de usar `exit()` para "encerrar".

</details>

````python
# Exercício 08 — Equação do 2º grau
import math

a = float(input("Valor de a: "))

if a == 0:
    print("Com a = 0 a equação não é do segundo grau.")
else:
    b = float(input("Valor de b: "))
    c = float(input("Valor de c: "))

    delta = (b ** 2) - (4 * a * c)
    print("Delta =", delta)

    if delta < 0:
        print("A equação não possui raízes reais.")
    elif delta == 0:
        x = -b / (2 * a)
        print("A equação possui uma raiz real:", x)
    else:
        x1 = (-b + math.sqrt(delta)) / (2 * a)
        x2 = (-b - math.sqrt(delta)) / (2 * a)
        print("A equação possui duas raízes reais:", x1, "e", x2)
````

## Exercício 09 🔴 — Folha de pagamento

Faça um programa para calcular uma folha de pagamento. Peça o **valor da hora** e a **quantidade de horas** trabalhadas no mês.

- O **INSS** desconta **10%** do salário bruto;
- O **FGTS** corresponde a **11%** do salário bruto, mas **não é descontado** (é a empresa que deposita);
- O **Imposto de Renda** depende do salário bruto:

| Salário Bruto | Desconto de IR |
|---|---|
| até 900 (inclusive) | isento (0%) |
| até 1500 (inclusive) | 5% |
| até 2500 (inclusive) | 10% |
| acima de 2500 | 20% |

Salário Líquido = Salário Bruto − (IR + INSS). Mostre as informações alinhadas, como no exemplo.

**Exemplos de execução:**

*Exemplo 1:*

```text
Valor da hora: R$ 5
Horas trabalhadas no mês: 220

Salário Bruto (5.0 * 220.0)     : R$   1100.00
(-) IR (5%)                     : R$     55.00
(-) INSS (10%)                  : R$    110.00
FGTS (11%)                      : R$    121.00
Total de descontos              : R$    165.00
Salário Líquido                 : R$    935.00
```

*Exemplo 2:*

```text
Valor da hora: R$ 30
Horas trabalhadas no mês: 160

Salário Bruto (30.0 * 160.0)    : R$   4800.00
(-) IR (20%)                    : R$    960.00
(-) INSS (10%)                  : R$    480.00
FGTS (11%)                      : R$    528.00
Total de descontos              : R$   1440.00
Salário Líquido                 : R$   3360.00
```

<details>
<summary>💡 <b>Dica</b> (clique para abrir)</summary>

Monte o rótulo e alinhe tudo com f-strings: `f'{texto:<32}: R$ {valor:9.2f}'` deixa o texto com 32 posições à esquerda e o número com 9 posições à direita.

</details>

````python
# Exercício 09 — Folha de pagamento
valor_hora = float(input("Valor da hora: R$ "))
horas = float(input("Horas trabalhadas no mês: "))

salario_bruto = valor_hora * horas

if salario_bruto <= 900:
    percentual_ir = 0
elif salario_bruto <= 1500:
    percentual_ir = 5
elif salario_bruto <= 2500:
    percentual_ir = 10
else:
    percentual_ir = 20

ir = salario_bruto * (percentual_ir / 100)
inss = salario_bruto * 0.10
fgts = salario_bruto * 0.11
total_descontos = ir + inss
salario_liquido = salario_bruto - total_descontos

print()
print(f"Salário Bruto ({valor_hora} * {horas}) : R$ {salario_bruto:.2f}")
print(f"(-) IR ({percentual_ir}%)                 : R$ {ir:.2f}")
print(f"(-) INSS (10%)              : R$ {inss:.2f}")
print(f"FGTS (11%)                  : R$ {fgts:.2f}")
print(f"Total de descontos          : R$ {total_descontos:.2f}")
print(f"Salário Líquido             : R$ {salario_liquido:.2f}")
````
