## Exercício 01 🟢 — Conversor de medidas

Faça um programa que peça uma medida em **metros** e mostre o valor equivalente em **centímetros** e em **milímetros**.

**Exemplo de execução:**

```text
Digite a medida em metros: 1.75
1.75 m equivalem a 175.0 cm e 1750.0 mm.
```

<details>
<summary>💡 <b>Dica</b> (clique para abrir)</summary>

1 metro = 100 centímetros = 1000 milímetros.

</details>

````python
# Exercício 01 — Conversor de medidas
metros = float(input("Digite a medida em metros: "))

centimetros = metros * 100
milimetros = metros * 1000

print(metros, "m equivalem a", centimetros, "cm e", milimetros, "mm.")
````

## Exercício 02 🟢 — Média bimestral

Faça um programa que peça as **4 notas bimestrais** de um aluno e mostre a **média** com duas casas decimais.

**Exemplo de execução:**

```text
Nota do 1º bimestre: 7
Nota do 2º bimestre: 8
Nota do 3º bimestre: 6.5
Nota do 4º bimestre: 9.5
Média final: 7.75
```

<details>
<summary>💡 <b>Dica</b> (clique para abrir)</summary>

Lembre-se dos parênteses: `(n1 + n2 + n3 + n4) / 4` é diferente de `n1 + n2 + n3 + n4 / 4`.

</details>

````python
# Exercício 02 — Média bimestral
nota1 = float(input("Nota do 1º bimestre: "))
nota2 = float(input("Nota do 2º bimestre: "))
nota3 = float(input("Nota do 3º bimestre: "))
nota4 = float(input("Nota do 4º bimestre: "))

media = (nota1 + nota2 + nota3 + nota4) / 4

print(f"Média final: {media:.2f}")
````

## Exercício 03 🟢 — Conversor de temperatura

Faça um programa que peça uma temperatura em graus **Celsius** e mostre o valor convertido para **Fahrenheit** e para **Kelvin**.

Fórmulas:

- `F = (C * 9/5) + 32`
- `K = C + 273.15`

**Exemplo de execução:**

```text
Temperatura em Celsius: 25
25.0 °C = 77.0 °F
25.0 °C = 298.15 K
```

````python
# Exercício 03 — Conversor de temperatura
celsius = float(input("Temperatura em Celsius: "))

fahrenheit = (celsius * 9 / 5) + 32
kelvin = celsius + 273.15

print(f"{celsius} °C = {fahrenheit:.1f} °F")
print(f"{celsius} °C = {kelvin:.2f} K")
````

## Exercício 04 🟡 — Salário com descontos

Faça um programa que pergunte quanto você ganha **por hora** e quantas **horas** trabalhou no mês. Sabendo que são descontados **11% de Imposto de Renda**, **8% de INSS** e **5% de sindicato**, mostre o resultado no formato abaixo:

```text
+ Salário Bruto   : R$
- IR (11%)        : R$
- INSS (8%)       : R$
- Sindicato (5%)  : R$
= Salário Líquido : R$
```

> Salário Líquido = Salário Bruto − descontos.

**Exemplo de execução:**

```text
Quanto você ganha por hora? R$ 25
Quantas horas trabalhou no mês? 160
+ Salário Bruto   : R$  4000.00
- IR (11%)        : R$   440.00
- INSS (8%)       : R$   320.00
- Sindicato (5%)  : R$   200.00
= Salário Líquido : R$  3040.00
```

<details>
<summary>💡 <b>Dica</b> (clique para abrir)</summary>

Para alinhar os números à direita, use largura na f-string: `f'{valor:8.2f}'` reserva 8 espaços com 2 casas decimais.

</details>

````python
# Exercício 04 — Salário com descontos
valor_hora = float(input("Quanto você ganha por hora? R$ "))
horas_mes = float(input("Quantas horas trabalhou no mês? "))

salario_bruto = valor_hora * horas_mes
ir = salario_bruto * 0.11
inss = salario_bruto * 0.08
sindicato = salario_bruto * 0.05
salario_liquido = salario_bruto - ir - inss - sindicato

print(f"+ Salário Bruto   : R$ {salario_bruto:.2f}")
print(f"- IR (11%)        : R$ {ir:.2f}")
print(f"- INSS (8%)       : R$ {inss:.2f}")
print(f"- Sindicato (5%)  : R$ {sindicato:.2f}")
print(f"= Salário Líquido : R$ {salario_liquido:.2f}")
````

## Exercício 05 🟡 — O pescador João

João, um pescador, precisa controlar o rendimento do seu trabalho. Sempre que ele traz mais de **50 kg** de peixe (limite do regulamento de pesca), paga uma multa de **R$ 4,00 por quilo excedente**.

Faça um programa que leia o **peso** dos peixes, guarde na variável `excesso` a quantidade de quilos acima do limite e na variável `multa` o valor a pagar. Mostre os resultados com mensagens adequadas.

Se João não passar do limite, o excesso e a multa devem ser **zero**.

**Exemplos de execução:**

*Exemplo 1:*

```text
Peso dos peixes (kg): 62.5
Excesso: 12.5 kg
Multa a pagar: R$ 50.00
```

*Exemplo 2:*

```text
Peso dos peixes (kg): 40
Excesso: 0 kg
Multa a pagar: R$ 0.00
```

<details>
<summary>💡 <b>Dica</b> (clique para abrir)</summary>

Ainda não estamos usando `if`! Pense em como a função `max()` pode evitar um excesso negativo: `max(0, -3)` → `0`.

</details>

````python
# Exercício 05 — O pescador João
peso = float(input("Peso dos peixes (kg): "))

excesso = max(0, peso - 50)
multa = excesso * 4

print("Excesso:", excesso, "kg")
print(f"Multa a pagar: R$ {multa:.2f}")
````

## Exercício 06 🟡 — Loja de tintas

Faça um programa para uma loja de tintas. O programa deve pedir o tamanho, em **metros quadrados**, da área a ser pintada. Considere que **1 litro** de tinta cobre **3 m²** e que a tinta é vendida apenas em **latas de 18 litros**, que custam **R$ 80,00**.

Informe a quantidade de **latas** a serem compradas e o **preço total**.

**Exemplo de execução:**

```text
Área a ser pintada (m²): 100
Você precisa de 33.33 litros de tinta.
Compre 2 lata(s) de 18 L.
Total: R$ 160.00
```

<details>
<summary>💡 <b>Dica</b> (clique para abrir)</summary>

Não dá para comprar meia lata! Use `math.ceil()` para arredondar para cima.

</details>

````python
# Exercício 06 — Loja de tintas
import math

area = float(input("Área a ser pintada (m²): "))

litros = area / 3
latas = math.ceil(litros / 18)
preco_total = latas * 80

print(f"Você precisa de {litros:.2f} litros de tinta.")
print("Compre", latas, "lata(s) de 18 L.")
print(f"Total: R$ {preco_total:.2f}")
````

## Exercício 07 🔴 — Loja de tintas — latas e galões

A loja de tintas agora vende dois tamanhos:

- **Lata de 18 litros** — R$ 80,00
- **Galão de 3,6 litros** — R$ 25,00

A cobertura é de **1 litro para cada 6 m²**, e você deve acrescentar **10% de folga** na quantidade de tinta. Peça a área a ser pintada e informe a quantidade e o preço em **3 situações**:

1. comprar **apenas latas** de 18 litros;
2. comprar **apenas galões** de 3,6 litros;
3. **misturar** latas e galões para desperdiçar menos tinta (use o máximo de latas cheias e complete com galões).

Sempre arredonde para cima (embalagens cheias). Mostre também quanto de tinta sobra em cada situação.

**Exemplo de execução:**

```text
Área a ser pintada (m²): 200
Tinta necessária (com 10% de folga): 36.67 L
1) Apenas latas : 3 lata(s) -> R$ 240.00 (sobra 17.33 L)
2) Apenas galões: 11 galão(ões) -> R$ 275.00 (sobra 2.93 L)
3) Misturando   : 2 lata(s) + 1 galão(ões) -> R$ 185.00 (sobra 2.93 L)
```

<details>
<summary>💡 <b>Dica</b> (clique para abrir)</summary>

Para a situação 3: `latas = litros // 18` dá quantas latas cabem *inteiras*. O que falta (`litros - latas * 18`) é coberto com galões, arredondando para cima.

</details>

````python
# Exercício 07 — Loja de tintas — latas e galões
import math

area = float(input("Área a ser pintada (m²): "))

litros = (area / 6) * 1.10

latas_so = math.ceil(litros / 18)
preco_latas = latas_so * 80
sobra_latas = latas_so * 18 - litros

galoes_so = math.ceil(litros / 3.6)
preco_galoes = galoes_so * 25
sobra_galoes = galoes_so * 3.6 - litros

latas_misto = int(litros // 18)
falta = litros - (latas_misto * 18)
galoes_misto = math.ceil(falta / 3.6)
preco_misto = (latas_misto * 80) + (galoes_misto * 25)
sobra_misto = (latas_misto * 18 + galoes_misto * 3.6) - litros

print(f"Tinta necessária (com 10% de folga): {litros:.2f} L")
print(f"1) Apenas latas : {latas_so} lata(s) -> R$ {preco_latas:.2f} (sobra {sobra_latas:.2f} L)")
print(f"2) Apenas galões: {galoes_so} galão(ões) -> R$ {preco_galoes:.2f} (sobra {sobra_galoes:.2f} L)")
print(f"3) Misturando   : {latas_misto} lata(s) + {galoes_misto} galão(ões) -> R$ {preco_misto:.2f} (sobra {sobra_misto:.2f} L)")
````

## Exercício 08 🔴 — Tempo de download detalhado

Faça um programa que peça o tamanho de um arquivo em **Gigabytes (GB)** e a velocidade da internet em **Mbps** (megabits por segundo). Calcule o tempo aproximado de download e mostre:

- o tamanho do arquivo em **MB**;
- o tempo total em **minutos** (2 casas decimais);
- o tempo no formato **horas, minutos e segundos** (ex.: `2h 16min 32s`).

Considere: **1 GB = 1024 MB** e **1 byte = 8 bits** (ou seja, 1 MB = 8 megabits).

**Exemplo de execução:**

```text
Tamanho do arquivo (GB): 50
Velocidade da internet (Mbps): 50
Tamanho: 51200 MB
Tempo aproximado: 136.53 minutos
Ou seja: 2h 16min 32s
```

<details>
<summary>💡 <b>Dica</b> (clique para abrir)</summary>

Calcule o total em segundos e converta para inteiro. Depois use `//` e `%`: `3725 // 3600` → `1` hora; `3725 % 3600` → `125` segundos restantes.

</details>

````python
# Exercício 08 — Tempo de download detalhado
tamanho_gb = float(input("Tamanho do arquivo (GB): "))
velocidade_mbps = float(input("Velocidade da internet (Mbps): "))

tamanho_mb = tamanho_gb * 1024
tamanho_megabits = tamanho_mb * 8
tempo_segundos = tamanho_megabits / velocidade_mbps
tempo_minutos = tempo_segundos / 60

total_seg = int(tempo_segundos)
horas = total_seg // 3600
resto = total_seg % 3600
minutos = resto // 60
segundos = resto % 60

print(f"Tamanho: {tamanho_mb:.0f} MB")
print(f"Tempo aproximado: {tempo_minutos:.2f} minutos")
print(f"Ou seja: {horas}h {minutos}min {segundos}s")
````

## Exercício 09 🔴 — Caixa eletrônico

Faça um programa para um caixa eletrônico. O programa deve pedir o **valor do saque** (um número inteiro) e informar **quantas notas de cada valor** serão entregues, usando sempre o **menor número de notas possível**.

As notas disponíveis são de **R$ 100, 50, 10, 5 e 1**. Não se preocupe com a quantidade de notas na máquina.

Mostre a quantidade de todas as notas, mesmo as que ficarem com zero.

**Exemplos de execução:**

*Exemplo 1:*

```text
Valor do saque: R$ 256
2 nota(s) de R$ 100
1 nota(s) de R$ 50
0 nota(s) de R$ 10
1 nota(s) de R$ 5
1 nota(s) de R$ 1
```

*Exemplo 2:*

```text
Valor do saque: R$ 399
3 nota(s) de R$ 100
1 nota(s) de R$ 50
4 nota(s) de R$ 10
1 nota(s) de R$ 5
4 nota(s) de R$ 1
```

<details>
<summary>💡 <b>Dica</b> (clique para abrir)</summary>

Comece pela nota maior. `256 // 100` → `2` notas de 100; `256 % 100` → `56` reais que ainda faltam entregar. Repita com a próxima nota.

</details>

````python
# Exercício 09 — Caixa eletrônico
valor = int(input("Valor do saque: R$ "))

notas_100 = valor // 100
resto = valor % 100

notas_50 = resto // 50
resto = resto % 50

notas_10 = resto // 10
resto = resto % 10

notas_5 = resto // 5
notas_1 = resto % 5

print(notas_100, "nota(s) de R$ 100")
print(notas_50, "nota(s) de R$ 50")
print(notas_10, "nota(s) de R$ 10")
print(notas_5, "nota(s) de R$ 5")
print(notas_1, "nota(s) de R$ 1")
````
